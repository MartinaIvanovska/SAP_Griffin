```python
import pandas as pd

df = pd.read_csv("../../creditcard.csv")
```

```python
df.head()
```

```python
len(df)
```

```python
normal = df[df["Class"] == 0]
fraud = df[df["Class"] == 1]
```

```python
len(normal)
```

```python
len(fraud)
```

```python
context_normal = normal.sample(n=1800, random_state=42)
context_fraud = fraud.sample(n=50, random_state=42)
```

```python
context = pd.concat(
    [context_normal, context_fraud]
)
```

```python
remaining = df.drop(context.index)
```

```python
context = context.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
remaining_normal = remaining[remaining["Class"] == 0]
remaining_fraud = remaining[remaining["Class"] == 1]
```

```python
normal_test = remaining_normal.sample(n=150, random_state=123)
fraud_test = remaining_fraud.sample(n=30, random_state=123)
```

```python
test = pd.concat(
    [normal_test, fraud_test]
).sample(frac=1, random_state=42)
```

```python
test["ActualClass"] = test["Class"]
```

```python
test["Class"] = "[PREDICT]"
```

```python
test.head()
```

```python
data = pd.concat(
    [context, test],
    ignore_index=True
)
```

```python
data
```

```python
data.to_csv("data_creditcard_with_class_more_testcases.csv", index=False)
```

```python
sap_test = test.drop(columns=["ActualClass"])
```

```python
data_test = pd.concat(
    [context, sap_test],
    ignore_index=True
)
```

```python
data_test.to_csv(
    "data_test_creditcard_more_testcases.csv",
    index=False
)
```

```python
import pandas as pd

df = pd.read_csv("creditcard.csv")
```

```python
df.head()
```

```python
normal = df[df["Class"] == 0]
fraud = df[df["Class"] == 1]
```

```python
context_normal = normal.sample(n=90, random_state=42)
context_fraud = fraud.sample(n=10, random_state=42)
```

```python
context = pd.concat(
    [context_normal, context_fraud]
)
```

```python
remaining = df.drop(context.index)
```

```python
test = remaining.sample(n=10, random_state=123).copy()
```

```python
test
```

```python
test["Class"] = "[PREDICT]"
```

```python
data = pd.concat(
    [context, test],
    ignore_index=True
)
```

```python
data
```

```python
data.to_csv("data_balanced_sample.csv", index=False)
```
